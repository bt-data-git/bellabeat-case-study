# Bellabeat Case Study — Analyze Phase

## INCREASING READABILITY
Changing the column format from minutes to hours for better comprehension of the values

```python
activity_all['SedentaryHours'] = activity_all['SedentaryMinutes'] / 60

activity_all['VeryActiveHours'] = activity_all['VeryActiveMinutes'] / 60
activity_all['FairlyActiveHours'] = activity_all['FairlyActiveMinutes'] / 60
activity_all['LightlyActiveHours'] = activity_all['LightlyActiveMinutes'] / 60
```

```python
sleep_clean['HoursAsleep'] = sleep_clean['TotalMinutesAsleep'] / 60
sleep_clean['HoursInBed'] = sleep_clean['TotalTimeInBed'] / 60
```

Summary statistics are calculated for key activity measures to understand the typical values, variation, and overall range within the dataset.

```python
activity_summary = activity_all[
    [
        'TotalSteps',
        'VeryActiveHours',
        'FairlyActiveHours',
        'LightlyActiveHours',
        'SedentaryHours',
        'Calories'
    ]
].describe().round(2)

activity_summary
```

## OVERALL SLEEP SUMMARY
Calculate summary statistics like count, mean, median, min, max, etc.

## TOTAL AWAKE IN BED TIME

```python
sleep_summary = sleep_clean[
    [
        'TotalSleepRecords',
        'HoursAsleep',
        'HoursInBed'
    ]
].describe().round(2)

sleep_summary
```

## TOTAL ACTIVE TIME

```python
sleep_clean['TimeAwakeInBedHours'] = (
    sleep_clean['HoursInBed'] - sleep_clean['HoursAsleep']
)

sleep_clean['TimeAwakeInBedHours'].describe().round(2)
```

## BETTER READING COMPREHENSION
For better reading comprehension, transformed duration format from decimals to words
TotalActiveHours -> numeric decimal (3.69)
TotalActiveTime -> formatted text (3h 41m)

```python
activity_all['TotalActiveMinutes'] = (
    activity_all['VeryActiveMinutes']
    + activity_all['FairlyActiveMinutes']
    + activity_all['LightlyActiveMinutes']
)

activity_all['TotalActiveHours'] = activity_all['TotalActiveMinutes'] / 60

activity_all['TotalActiveHours'].describe().round(2)
```

```python
def hours_to_hm(hours):
    total_minutes = round(hours * 60)
    h = total_minutes // 60
    m = total_minutes % 60
    return f"{h}h {m}m"
```

```python
activity_all['TotalActiveTime'] = activity_all['TotalActiveHours'].apply(hours_to_hm)

sleep_clean['SleepDuration'] = sleep_clean['HoursAsleep'].apply(hours_to_hm)
sleep_clean['TimeInBed'] = sleep_clean['HoursInBed'].apply(hours_to_hm)
sleep_clean['AwakeInBedTime'] = sleep_clean['TimeAwakeInBedHours'].apply(hours_to_hm)
```

```python
activity_all[
    ['TotalActiveHours', 'TotalActiveTime']
].head()
```

## WEEKDAY PATTERNS - CREATING COLUMNS
Creating weekday columns

```python
activity_all['DayOfWeek'] = activity_all['ActivityDate'].dt.day_name()

sleep_clean['DayOfWeek'] = sleep_clean['SleepDay'].dt.day_name()
```

## WEEKDAY ACTIVITY ANALYSIS
Calculate average steps, active hours, sedentary hours, and calories by day of week to identify weekly activity patterns.

```python
activity_by_day = activity_all.groupby('DayOfWeek')[
    [
        'TotalSteps',
        'TotalActiveHours',
        'SedentaryHours',
        'Calories'
    ]
].mean().round(2)

activity_by_day
```

```python
sleep_by_day = sleep_clean.groupby('DayOfWeek')[
    [
        'HoursAsleep',
        'HoursInBed',
        'TimeAwakeInBedHours'
    ]
].mean().round(2)

sleep_by_day
```

## REORDERING WEEKDAYS
Pandas grouped weekday names alphabetically, so I reorder the index to Monday–Sunday for clearer interpretation.

```python
day_order = [
    'Monday',
    'Tuesday',
    'Wednesday',
    'Thursday',
    'Friday',
    'Saturday',
    'Sunday'
]
```

```python
activity_by_day = activity_by_day.reindex(day_order)
sleep_by_day = sleep_by_day.reindex(day_order)
```

```python
activity_by_day
```

```python
sleep_by_day
```

## WEEKDAY FINDINGS
Activity levels were highest on Saturday and lowest on Sunday. Sleep duration peaked on Sunday, while Thursday showed the shortest average sleep. This suggests user behaviour varies by day of week, with weekends showing different activity and sleep patterns from weekdays.

## SLEEP/ACTIVITY RELATIONSHIP ANALYSIS
Using correlation to see whether things like more steps or more active time are associated with more sleep.

```python
sleep_activity_corr = activity_sleep[
    [
        'TotalSteps',
        'TotalActiveHours',
        'SedentaryHours',
        'Calories',
        'HoursAsleep',
        'HoursInBed',
        'TimeAwakeInBedHours'
    ]
].corr().round(2)

sleep_activity_corr
```

## CORRELATION FINDINGS
Total steps and total active hours showed little relationship with sleep duration (r = -0.19 and r = -0.07, respectively). This was somewhat unexpected, as greater daily activity might reasonably be expected to correspond with longer sleep. Sedentary hours showed a stronger negative relationship with sleep duration (r = -0.60), although this should be interpreted cautiously because both measures represent portions of a finite 24-hour day. These correlations describe associations only and do not demonstrate causation.

## USER-LEVEL ACTIVITY/SLEEP SUMMARY
What does an average user look like ?

```python
user_activity_summary = activity_all.groupby('Id').agg(
    AvgSteps=('TotalSteps', 'mean'), 
    AvgActiveHours=('TotalActiveHours', 'mean'),
    AvgSedentaryHours=('SedentaryHours', 'mean'),
    AvgCalories=('Calories', 'mean'),
    DaysRecorded=('ActivityDate', 'count')
).round(2)

user_activity_summary.head()
```

```python
user_sleep_summary = activity_sleep.groupby('Id').agg(
    AvgSteps=('TotalSteps', 'mean'),
    AvgActiveHours=('TotalActiveHours', 'mean'),
    AvgSedentaryHours=('SedentaryHours', 'mean'),
    AvgSleepHours=('HoursAsleep', 'mean'),
    SleepDays=('ActivityDate', 'count')
).round(2)

user_sleep_summary.head()
```

## USER-LEVEL ACTIVITY/SLEEP FINDINGS:
User-level participation and behaviour varied considerably. Some participants contributed substantially more activity records than others, while average activity levels also differed between users. Sleep coverage was particularly uneven, with some users contributing only 3–5 sleep records while others contributed more than 25.
